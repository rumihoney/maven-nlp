```python
# importing spacy 
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
# view a single phrase
phrase = 'im selling lemons for $5 today'
phrase
```

```python
# create a spacy doc object
doc = nlp(phrase)
doc
```

```python
# Tokenization
[token.text for token in doc]
```

```python
# Lemmatization
[token.lemma_ for token in doc]
```

```python
# Removing stop words
norm = [token.lemma_ for token in doc if not token.is_stop]
norm 
```

```python
' '.join(norm)
```

```python
# View parts of speech tagging
pos = [(token.text, token.pos_) for token in doc]
pos
```

```python
# function to do all three things 
def token_lemma_nonstop(text):
    doc = nlp(text)
    output = [token.lemma_ for token in doc if not token.is_stop] # tokenize, lemmatize & removing stop words
    output = ' '.join(output)
    return output 
```

```python
# original sentence
phrase
```

```python
# applying the function on it
token_lemma_nonstop(phrase)
```
