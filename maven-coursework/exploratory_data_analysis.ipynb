## Filtering

```python
import pandas as pd
```

```python
# reading the data that was previously cleaned
groceries = pd.read_pickle('groceries_with_new_columns.pkl')
groceries.head()
```

```python
# filtering through 'low inventory' section
groceries.loc[groceries['Low Inventory'] == 'Low Inventory', ['Price_Dollars', 'Inventory']]
```

```python
# filtering through the 'subcategory' section
groceries[groceries.Subcategory == 'Snacks']
# output is empty because it is inputed with a space ' Fruit'
```

```python
groceries.Subcategory[0]
```

```python
# remove the space: ' Fruit' -> 'Fruit'
groceries.Subcategory = groceries.Subcategory.str.strip()
```

```python
# filtering with multiple conditions
mask = (groceries['Low Inventory'] == 'Low Inventory') & (groceries.Subcategory == 'Snacks')
```

```python
# this will show columns that both have Low Inventory and Snacks
groceries[mask]
```

```python
# filtering w multiple conditions and columns
groceries.loc[mask, ['Subcategory', 'Item', 'Inventory']]
```
