## Filtering

```python
import pandas as pd
```

```python
# reading the data that was previously cleaned
groceries = pd.read_pickle('groceries_with_new_columns.pkl')
groceries.head()
```

```python
# filtering through 'low inventory' section
groceries.loc[groceries['Low Inventory'] == 'Low Inventory', ['Price_Dollars', 'Inventory']]
```

```python
# filtering through the 'subcategory' section
groceries[groceries.Subcategory == 'Snacks']
# output is empty because it is inputed with a space ' Fruit'
```

```python
groceries.Subcategory[0]
```

```python
# remove the space: ' Fruit' -> 'Fruit'
groceries.Subcategory = groceries.Subcategory.str.strip()
```

```python
# filtering with multiple conditions
mask = (groceries['Low Inventory'] == 'Low Inventory') & (groceries.Subcategory == 'Snacks')
```

```python
# this will show columns that both have Low Inventory and Snacks
groceries[mask]
```

```python
# filtering w multiple conditions and columns
groceries.loc[mask, ['Subcategory', 'Item', 'Inventory']]
```

## Sorting

```python
groceries.head()
```

```python
# prices in ascending order (how to high)
prices_lowtohigh = groceries.sort_values('Price_Dollars')
prices_lowtohigh.head()
```

```python
# prices in descending order (high to low)
prices_hightolow = groceries.sort_values('Price_Dollars', ascending=False)
prices_hightolow.head()
```

```python
# sort prices within each category 
# subcatories are now alphabetical and the prices are also low to high
groceries.sort_values(['Subcategory', 'Price_Dollars'])
```

```python
# chain on a sort
groceries[groceries['Low Inventory'] == 'Low Inventory'].sort_values('Price_Dollars')
```

## Grouping

```python
groceries.head()
```

```python
# group by categories: total inventory
groceries.groupby('Category')['Inventory'].sum()
```

```python
# group by categories: number of rows
groceries.groupby('Category')['Inventory'].count()
```

```python
groceries.groupby(['Category', 'Subcategory'])['Inventory'].sum()
```

```python
# changing it back to a dataframe with '.reset_index()' at the end 
groceries.groupby(['Category', 'Subcategory'])['Inventory'].sum().reset_index()
```

```python
# adding aggregations to look at the 'sum' and 'count' 
groceries.groupby(['Category', 'Subcategory'])['Inventory'].agg(['sum', 'count']).reset_index()
```

```python
# other way to write it with added parentheses ()
(groceries.groupby(['Category', 'Subcategory'])['Inventory']
 .agg(['sum', 'count'])
 .reset_index())
```

```python
# max price within each category
groceries.groupby('Category')['Price_Dollars'].max()
```

```python
# item associated with the max prices
(groceries[['Category', 'Item', 'Price_Dollars']]
        .sort_values('Price_Dollars', ascending=False)
        .groupby('Category')
        .head(1))
```

## Visualizing Data

### Plots available with pandas:
1. **Line:** `.plot.line()` _Trends over time or ordered intervals_
2. **Vertical Bar:** `.plot.bar()` _Comparisons across discrete categories_
3. **Hortizontal Bar:** `.plot.barh()` _Comparisons with long category names_
4. **Histogram:** `.plot.hist()` _Data distribution and frequencies_
5. **Box Plot:** `.plot.box()` _Visualizing medians, outliers, and quartiles_
6. **Scatter:** `.plot.scatter(x,y)` _Relationships between 2 variables (DataFrame only)_
7. **Area:** `.plot.area()` _Cumulative totals over time_
8. **Pie:** `.plot.pie(y=...)` _Portortions of a whole_

There are many more available here: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.html

Use *.plot* when creating quick visualization from a DataFrame, and *matplotlib* or/and *seaborn* for fine tuning visualizations with more complex data

## Bar plot

```python
# vertical plot: subcategory & price average plot
groceries.groupby(['Subcategory'])['Price_Dollars'].mean().plot.bar();
```

```python
# horizontal plot: barh
groceries.groupby(['Subcategory'])['Price_Dollars'].mean().plot.barh();
```

```python
# sorting values
groceries.groupby(['Subcategory'])['Price_Dollars'].mean().sort_values()
```

```python
# visualization of sorted values
groceries.groupby(['Subcategory'])['Price_Dollars'].mean().sort_values().plot.barh();
```

## Line plot

```python
happiness = pd.read_csv('happiness_survey_data.csv')
happiness.head()
```

```python
happiness[happiness.country_name == 'Mexico'].iloc[:, 1:3].plot.line(x='year', y='happiness_score')
```

```python
# line chart for Canada, Mexico and USA
(happiness[happiness.country_name
           .isin(['Canada', 'Mexico', 'United States'])]
           .iloc[:, :3]
           .pivot(index='year', columns='country_name', values='happiness_score')
           .plot.line())
```
