## Reading CSV and Excel files with Pandas

```python
import os
print(os.getcwd())
```

```python
import pandas as pd
```

```python
pd.read_csv('course_offerings.csv')
```

```python
# adding more arguments -> separator ',' and header
pd.read_csv('course_offerings.csv', sep=',', header='infer')
```

```python
# reading the second tab of an excel file
pd.read_excel('Course Offerings.xlsx', sheet_name=1)
```

```python
# connect to SQLite database
import sqlite3
conn = sqlite3.connect('../Data/maven.db') 
```

```python
pd.read_sql('SELECT * FROM courses', conn)
```

```python
pd.read_sql('SELECT * FROM courses WHERE course_id=101', conn)
```
